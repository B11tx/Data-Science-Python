```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "amount": [500, 1500, 2000, 300],
    "country": ["PL", "PL", "DE", "PL"],
    "discount": [0.1, 0.2, 0.0, 0.15]
})


df["final_price"] = np.where(
    (df["country"] == "PL") & (df["amount"] > 1000),
    df["amount"] * (1 - df["discount"]),
    df["amount"]
)

print(df)
```
