```python
import matplotlib.pyplot as plt

# Konfiguracja danych
# Dni od 1 do 31
rows = [f"{i}." for i in range(1, 32)]
# Dodajemy wiersz podsumowania na dole
rows.append("RAZEM")

# Nagłówki: Data + 6 pustych miejsc na imiona
columns = ["Data"] + [" " for _ in range(6)]

# Puste dane w środku (puste ciągi znaków)
cell_text = [["" for _ in range(7)] for _ in range(32)]

# Wypełnienie pierwszej kolumny numerami dni
for i in range(31):
    cell_text[i][0] = rows[i]

# Wypełnienie ostatniego wiersza (RAZEM)
cell_text[31][0] = "RAZEM"

# Ustawienia wykresu (format zbliżony do kartki A4 w pionie, jak na zdjęciu)
fig, ax = plt.subplots(figsize=(8.27, 11.69)) # A4 w calach
ax.axis('tight')
ax.axis('off')

# Tworzenie tabeli
table = ax.table(cellText=cell_text,
                 colLabels=columns,
                 loc='center',
                 cellLoc='center',
                 colWidths=[0.1] + [0.15]*6) # Data węższa, pracownicy szersi

# Stylizacja tabeli
table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1, 1.8) # Rozciągnięcie wierszy w pionie (żeby było miejsce na wpisanie godzin)

# Modyfikacja wyglądu komórek
for (row, col), cell in table.get_celld().items():
    # Ustawienie wysokości nagłówka
    if row == 0:
        cell.set_height(0.05)
        cell.set_text_props(weight='bold')
        # Jeśli to nagłówki pracowników (kolumny 1-6), dodaj linię do wpisania imienia
        if col > 0:
             cell.get_text().set_text("\n")

    # Stylizacja kolumny z datą (pogrubienie)
    if col == 0 and row > 0:
        cell.set_text_props(weight='bold')

    # Stylizacja wiersza RAZEM
    if row == 32:
        cell.set_text_props(weight='bold')


plt.show()
```
