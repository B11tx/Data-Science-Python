```python
import pandas as pd

df = pd.read_csv("housing.csv")

df.head()

print("\n Laczna pamiec ")
df.info(memory_usage="deep")

print("\n Pamiec dla kazdej kolumny ")
df.memory_usage(deep=True)
```

· Które kolumny zużywają najwięcej pamięci?

Najwięcej zużywa kolumna: ocean_proximity: 1 177 820 bajtów.
Pozostałe kolumny:longitude, latitude, population: zużywają po 165 120 bajtów

· Czy to głównie liczby, czy stringi (object)?

Większość danych to liczby(float), ale to stringi (typ object) generują największe zużycie pamięci.
dtypes: float64(9), object(1)
