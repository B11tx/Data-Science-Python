```python
import pandas as pd

def licz_sume_naivnie(path, kolumna, wielkosc_fragmentu=5000):
    total_sum = 0

    for chunk in pd.read_csv(path, chunksize=wielkosc_fragmentu):

        total_sum += chunk[kolumna].sum()
    return total_sum


wynik = licz_sume_naivnie("housing.csv", "total_rooms")
print(f"Suma pokoi (total_rooms): {wynik}")
```
